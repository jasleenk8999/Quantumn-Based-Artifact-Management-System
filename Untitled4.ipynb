```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import MinMaxScaler
from sklearn.model_selection import train_test_split
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense
```

```python
files = [
    "train_FD001.txt",
    "train_FD002.txt",
    "train_FD003.txt",
    "train_FD004.txt"
]

df_list = []
engine_offset = 0

for file in files:
    temp_df = pd.read_csv(file, sep=" ", header=None)
    temp_df = temp_df.dropna(axis=1)

    # Fix engine_id overlap
    temp_df[0] = temp_df[0] + engine_offset
    engine_offset = temp_df[0].max()

    df_list.append(temp_df)

# Combine all datasets
df = pd.concat(df_list, ignore_index=True)
```

```python
cols = ['engine_id', 'cycle'] + \
       [f'op{i}' for i in range(1,4)] + \
       [f's{i}' for i in range(1,22)]

df.columns = cols

df.head()
```

```python
max_cycle = df.groupby('engine_id')['cycle'].max()

df['RUL'] = df.apply(lambda row: max_cycle[row['engine_id']] - row['cycle'], axis=1)

# Normalize RUL → health (0 to 1)
max_rul = df['RUL'].max()
df['health'] = df['RUL'] / max_rul
```

```python
features = ['s11','s12','s13','s15']

scaler = MinMaxScaler()
df[features] = scaler.fit_transform(df[features])
```

```python
def create_sequences(df, window=3):
    X, y = [], []

    for engine in df['engine_id'].unique():
        engine_df = df[df['engine_id'] == engine]
        data = engine_df[features].values
        target = engine_df['health'].values

        for i in range(len(data) - window):
            X.append(data[i:i+window])
            y.append(target[i+window])

    return np.array(X), np.array(y)

X, y = create_sequences(df, window=3)

print("Shape of X:", X.shape)   # (?, 3, 4)
print("Shape of y:", y.shape)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, shuffle=True
)
```

```python
model = Sequential([
    LSTM(64, input_shape=(X.shape[1], X.shape[2])),
    Dense(32, activation='relu'),
    Dense(1)
])

model.compile(optimizer='adam', loss='mse')
```

```python
model.fit(X_train, y_train, epochs=10, batch_size=32)
```

```python
loss = model.evaluate(X_test, y_test)
print("Test Loss:", loss)
```
